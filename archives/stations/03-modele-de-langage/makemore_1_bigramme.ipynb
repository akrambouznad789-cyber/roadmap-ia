# makemore 1 : le modèle bigramme

Suivi de [The spelled-out intro to language modeling: building makemore](https://www.youtube.com/watch?v=PaCmpygFfXo) (Andrej Karpathy).

Données : `donnees/names.txt`, 32 032 prénoms, tirés du dépôt [makemore](https://github.com/karpathy/makemore) (licence MIT). Quand Karpathy écrit `open('names.txt', 'r')`, écris `open('donnees/names.txt', 'r')`.

```python
words = open('donnees/names.txt', 'r').read().splitlines()
```

```python
words[:10]
```

```python
print(len(words),
min(len(w) for w in words),
max(len(w) for w in words))
```

```python
b =  {}
# Bigrams 'pair of letters combinations'
for w in words:
    chs = ['<S>'] + list(w) + ['<E>']
    for ch1, ch2 in zip(chs, chs[1:]):
        bigram = (ch1, ch2)
        b[bigram] = b.get(bigram, 0) + 1
```

```python
sorted(b.items(), key = lambda kv: -kv[1])
```

```python
#storing in a 2D array, w/ PyTorch
import torch
```

```python
N = torch.zeros ((28,28), dtype = torch.int32)
```

```python
chars = sorted(list(set(''.join(words))))
stoi =  {s:i for i,s in enumerate(chars)}
stoi['<S>'] = 26
stoi['<E>'] = 27
```

```python
for w in words:
    chs = ['<S>'] + list(w) + ['<E>']
    for ch1, ch2 in zip(chs, chs[1:]):
        ix1 = stoi[ch1]
        ix2 = stoi[ch2]
        N[ix1, ix2] += 1
```

```python
itos = {i:s for s,i in stoi.items()}
```
